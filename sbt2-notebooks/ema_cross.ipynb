# EMA cross

Runs `ema_cross.py` on the sbt2 server. Set `SBT2_SERVER_TOKEN`, and `SBT2_SERVER_URL` if the server is not on this machine.

```python
import os

from sbt2.lab import Lab

url = os.environ.get("SBT2_SERVER_URL", "ws://127.0.0.1:8765")
lab = await Lab.connect(url, os.environ["SBT2_SERVER_TOKEN"])
```

```python
runs = await lab.run(
    {
        "instruments": ["BTCUSDT-LINEAR.BYBIT"],
        "period": ["2025-01-01", "2025-01-31"],
        "venue": "bybit_linear",
        "capital": "10000 USDT",
        "bars": "candles",
        "split": {"validation": 0.2, "test": 0.2},
        "params": {"fast": [6, 12], "slow": 48},
    },
    "ema_cross:EmaCross",
)
runs
```

```python
await runs[0].tearsheet()
```

```python
await runs[0].metrics()
```

```python
await lab.close()
```
